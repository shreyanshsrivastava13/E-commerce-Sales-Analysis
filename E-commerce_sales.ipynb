```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.colors as colors
pio.templates.default = "plotly_white"
```

```python
data = pd.read_csv("Sample - Superstore.csv" , encoding = "latin-1")
data.head()
```

```python
data.describe()
```

```python
data.info()
```

# Coverting the "Order date" and "Ship date" columns to datetime format

```python
#data['Order Date'] = pd.to_datetime(data['Order Date']) .. it is convert only one column to datetime format. 
#it is convert multiple columns to datetime format.
data[['Order Date', 'Ship Date']] = data[['Order Date', 'Ship Date']].apply(pd.to_datetime)
data.info()
```

```python
data.head()
```

# Add 3 new columns to the dataset: 'Year', 'Month', and 'Week' 


```python
data["Order Year"] = data["Order Date"].dt.year
data["Order Month"] = data["Order Date"].dt.month
data["Order Day of Week"] = data["Order Date"].dt.dayofweek
```

```python
data.head()
```

# Monthly Sales Analysis

```python
sales_by_month = data.groupby("Order Month")["Sales"].sum().reset_index()
sales_by_month
```

```python
fig = px.line(sales_by_month, 
              x="Order Month", 
              y="Sales",
              title="Sales by Month",
)
fig.show()
```

# Sales by Category 


```python
sales_by_category = data.groupby("Category")["Sales"].sum().reset_index()
sales_by_category
```

```python
fig = px.pie(sales_by_category,
                values="Sales", 
                names="Category",
                hole=0.5,
                color_discrete_sequence=px.colors.sequential.RdBu,
)
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.update_layout(title_text="Sales by Category" , title_font = dict(size=20, family='Arial', color='black'))

fig.show()
```

# Sales Analyse by Sub category

```python
sales_by_sub_category = data.groupby("Sub-Category")["Sales"].sum().reset_index()
sales_by_sub_category
```

```python
fig = px.bar(sales_by_sub_category,
                x="Sub-Category", 
                y="Sales",
                title="Sales by Sub-Category",
                color="Sales",
                color_continuous_scale=px.colors.sequential.RdBu,
    )

fig.show()
```

# Monthly Profit Analysis

```python
profit_by_month = data.groupby("Order Month")["Profit"].sum().reset_index()
profit_by_month
```

```python
fig = px.bar(profit_by_month,
                x="Order Month", 
                y="Profit",
                title="Profit by Month",
                color="Profit",
                color_continuous_scale=px.colors.sequential.RdBu,
    )
fig.show()
```

# Profit analyse by Category

```python
profit_by_category = data.groupby("Category")["Profit"].sum().reset_index()
profit_by_category
```

```python
fig = px.treemap(profit_by_category,
                path=["Category"], 
                values="Profit",
                title="Profit by Category",
                color="Profit",
                color_continuous_scale=px.colors.sequential.RdBu,
    )
fig.show()
```

# Profit Analyse by Sub Category

```python
profit_by_sub_category = data.groupby("Sub-Category")["Profit"].sum().reset_index()
profit_by_sub_category
```

```python
fig = px.bar(profit_by_sub_category,
                x="Profit", 
                y="Sub-Category",
                orientation='h',
                title="Profit by Sub-Category",
                color="Profit",
                color_continuous_scale="RdYlGn",
    )
fig.show()
```

# Sales and Profit for Customer Segment 

```python
data.head()
```

```python
sales_profit_by_segment = data.groupby("Segment").agg({"Sales": "sum", "Profit": "sum"}).reset_index()
sales_profit_by_segment
```

```python
color_palette = colors.qualitative.Pastel

fig = go.Figure()
fig.add_trace(go.Bar(
    x=sales_profit_by_segment["Segment"],
    y=sales_profit_by_segment["Sales"],
    name="Sales",
    marker_color=color_palette[0]
))

fig.add_trace(go.Bar(
    x=sales_profit_by_segment["Segment"],
    y=sales_profit_by_segment["Profit"],
    name="Profit",
    marker_color=color_palette[1]
))

fig.update_layout(
    title="Sales and Profit by Customer Segment",
    xaxis_title="Customer Segment",
    yaxis_title="Amount",
)

fig.show()
```

# Profit Ratio by Sales Analyse

```python
sales_profit_by_segment = data.groupby("Segment")[["Sales", "Profit"]].sum().reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales'] / sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment', 'Sales_to_Profit_Ratio']])
```
